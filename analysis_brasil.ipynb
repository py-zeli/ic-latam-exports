```python
import pandas as pd
```

Ler fontes de dados oficiais

```python
df_exp = pd.read_csv('databases/govbr/EXP_2025.csv', delimiter=';',encoding='cp1252')
```

```python
df_ncm = pd.read_csv('databases/govbr/NCM.csv', delimiter=';', encoding='cp1252')
```

```python
df_pais = pd.read_csv('databases/govbr/PAIS.csv', delimiter=';', encoding='cp1252')
```

Gerar tabela unificada

```python
DfExportacaoNCM = pd.merge(df_exp, df_ncm[['CO_NCM', 'NO_NCM_POR']],on= 'CO_NCM')
```

```python
DfExportacaoNCMporPais = pd.merge(DfExportacaoNCM, df_pais[['CO_PAIS', 'NO_PAIS']],on= 'CO_PAIS')
```

Análise de Principais produtos exportados do Brasil

```python
DfExportacaoValorNCMAgregado = DfExportacaoNCM.groupby(['CO_NCM', 'NO_NCM_POR'])[['VL_FOB']].sum().reset_index()
```

```python
TotalExportacaoBrasil = DfExportacaoValorNCMAgregado['VL_FOB'].sum()
```

```python
DfExportacaoValorNCMAgregado['PCT_TOTAL'] = DfExportacaoValorNCMAgregado.apply(lambda row: (row['VL_FOB'] / TotalExportacaoBrasil) * 100, axis=1)
```

```python
DfExportacaoValorNCMAgregado.sort_values('VL_FOB', ascending=False).head(10)
```

Análise de Principais Destinos de Exportação

```python
DfExportacaoNCMporPais = pd.merge(DfExportacaoNCM, df_pais[['CO_PAIS', 'NO_PAIS']],on= 'CO_PAIS')
```

```python
RankingPaisesExportacao = DfExportacaoNCMporPais.groupby('NO_PAIS')['VL_FOB'].sum().reset_index()
```

```python
RankingPaisesExportacao.sort_values('VL_FOB', ascending=False).head(10)
```

Estudo sobre exportações à Argentina

```python
df_filtrado = DfExportacaoNCMporPais.loc[DfExportacaoNCMporPais['NO_PAIS'] == 'Argentina']
```

```python
df_filtrado = df_filtrado.groupby(['CO_NCM', 'NO_NCM_POR'])['VL_FOB'].sum().reset_index()
```

```python
df_filtrado.sort_values('VL_FOB', ascending=False).head(10)
```
