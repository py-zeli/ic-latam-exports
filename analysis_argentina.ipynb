### Análise dos dados argentinos ###

```python
import pandas as pd
```

```python
df_ncm = pd.read_csv('databases/govbr/NCM.csv', delimiter=';', encoding='cp1252')
```

```python
ArgentinaEXP_IMP = pd.read_csv('databases/secem_mercosur/NCM_2026-08-15.csv')
```

```python
ArgentinaEXP_IMP[["Exportaciones", "Importaciones (FOB)", "Importaciones (CIF)"]] = ArgentinaEXP_IMP[["Exportaciones", "Importaciones (FOB)", "Importaciones (CIF)"]].astype(int)
```

```python
ArgentinaEXP_IMP.rename(columns={'NCM':'CO_NCM'}, inplace=True)
```

```python
ArgentinaEXP_IMP = ArgentinaEXP_IMP.merge(df_ncm[['CO_NCM', 'NO_NCM_POR']], on = 'CO_NCM', how='left')
```

```python
ArgentinaEXP_IMP['NO_NCM_POR'] = ArgentinaEXP_IMP[
    'NO_NCM_POR'
].fillna(ArgentinaEXP_IMP['Descripción'])
```

```python
ArgentinaEXP_IMP = ArgentinaEXP_IMP[['CO_NCM', 'NO_NCM_POR', 'Exportaciones']]
```

```python
ArgentinaEXP_IMP = ArgentinaEXP_IMP.sort_values('Exportaciones', ascending=False)
```

```python
ArgentinaEXP_IMP['Exportaciones'].sum()
```

### Análise Importações por país

```python
ArgentinaImportNCM = pd.read_csv('databases/secem_mercosur/imports_by_country.csv', delimiter=';', encoding='utf-8')
```

```python
ArgentinaImportNCM['CIF(USD)'] = pd.to_numeric(
    ArgentinaImportNCM['CIF(USD)'].str.replace(',', '.', regex=False), errors='coerce'
).astype(int)
```

```python
ArgentinaImportNCM.rename(columns={'NCM':'CO_NCM'}, inplace=True)
```

```python
ArgentinaImportNCM = ArgentinaImportNCM.merge(df_ncm[['CO_NCM', 'NO_NCM_POR']], on = 'CO_NCM', how='left')
```

```python
ArgentinaImportNCM = ArgentinaImportNCM[['CO_NCM', 'NO_NCM_POR', 'Pais', 'CIF(USD)']]
```

```python
ArgentinaImportNCMAgregado = ArgentinaImportNCM.groupby(['CO_NCM', 'NO_NCM_POR','Pais'])['CIF(USD)'].sum().reset_index()
```

```python
ArgentinaImportNCMAgregado['CIF(USD)'].sum()
```

```python
ArgentinaImportNCMAgregado.sort_values('CIF(USD)', ascending=False, inplace=True)
```

```python
ArgentinaImportNCMAgregado.head(15)
```

```python
ArgentinaImportNCM.groupby(['CO_NCM', 'NO_NCM_POR'])['CIF(USD)'].sum().reset_index().sort_values('CIF(USD)', ascending=False).head(15)
```

```python
ArgentinaImportNCM.groupby('Pais')['CIF(USD)'].sum().reset_index().sort_values('CIF(USD)', ascending=False).head(15)
```
